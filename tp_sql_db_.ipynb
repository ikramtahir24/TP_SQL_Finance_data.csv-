```python
import pandas as pd
from pathlib import Path
from google.colab import drive

# Connexion à Google Drive
drive.mount('/content/drive')

# Mise à jour vers un chemin valide détecté dans votre Drive
chemin = Path('/content/drive/MyDrive/Colab Datasets/train.csv')

if not chemin.is_file():
    print("❌ Erreur : Le fichier spécifié n'a pas été trouvé.")
    mydrive_path = Path('/content/drive/MyDrive')
    if mydrive_path.exists():
        print("\nVoici les dossiers/fichiers disponibles à la racine de votre Drive :")
        for item in sorted(mydrive_path.iterdir()):
            if not item.name.startswith('.'):
                print(f" - {item.name}/" if item.is_dir() else f" - {item.name}")
    raise FileNotFoundError(f'Fichier introuvable : {chemin}')
else:
    df = pd.read_csv(chemin)
    display(df.head(10))
    print('Nombre de lignes :', df.shape[0])
    print('Nombre de colonnes :', df.shape[1])
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('tp_sql.db')

# Importer le DataFrame dans la table "donnees"
df.to_sql('donnees', conn, if_exists='replace', index=False)

# Première requête
resultat = pd.read_sql_query('SELECT * FROM donnees LIMIT 10;', conn)
display(resultat)

# Vérifier le nombre de lignes importées
total = pd.read_sql_query('SELECT COUNT(*) AS total FROM donnees;', conn)
display(total)

conn.close()

# Pour d'autres requêtes, rouvrez la connexion (la cellule ci-dessus la ferme à la fin) :
# conn = sqlite3.connect('tp_sql.db')
# pd.read_sql_query('SELECT * FROM donnees WHERE colonne = "valeur";', conn)
```
